```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression , LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split , GridSearchCV
from sklearn.metrics import accuracy_score , r2_score , mean_squared_error , precision_score , recall_score , classification_report , confusion_matrix

df_mat = pd.read_csv('student-mat.csv' , sep=';')
df_por = pd.read_csv('student-por.csv' , sep=';')

datasets = {'Mathematics': df_mat, 'Portuguese': df_por}

# Print both
for name, d in datasets.items():
  print(
      f'{name} Dataset -> Rows: {d.shape[0]}, Mean G3: {d["G3"].mean():.2f}'
  )

# Pick the one with the maximum row count
best_name = max(datasets, key=lambda k: datasets[k].shape[0])
df = datasets[best_name]

print(
    f'\n[RESULT] {best_name} dataset selected as MORE SUITABLE ({df.shape[0]}'
    ' rows).\n'
)

print(df.isnull().sum().sum())

df_encoded = pd.get_dummies(df, drop_first=True)



# ==========================================
# STEP 2: REMOVE OUTLIERS USING IQR METHOD (Lab 1)
# ==========================================
numeric_cols = ['G3', 'absences', 'studytime']
for col in numeric_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print(f'Dataset shape after removing outliers: {df.shape}')


x = df_encoded.drop(columns = 'G3')
y = df_encoded['G3']



scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

x_train , x_test , y_train , y_test = train_test_split(x_scaled , y , test_size=0.2 , random_state=42)

lir = LinearRegression()
lir.fit(x_train , y_train)
lir_pred = lir.predict(x_test)

mse = mean_squared_error(y_test, lir_pred)
r2 = r2_score(y_test, lir_pred)
print(f'Linear Regression MSE: {mse:.4f}')
print(f'Linear Regression R2 Score: {r2:.4f}')



y_class = (df_encoded['G3'] >= 10).astype(int)
X_class = df_encoded.drop(columns=['G3'])

scaler_class = StandardScaler()
X_scaled_c = scaler_class.fit_transform(X_class)

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_scaled_c, y_class, test_size=0.2, random_state=42, stratify=y_class
)

lor = LogisticRegression(random_state=42)
lor.fit(X_train_c, y_train_c)
y_pred_log = lor.predict(X_test_c)

dt_clf = DecisionTreeClassifier(criterion='gini', max_depth=5, random_state=42)
dt_clf.fit(X_train_c, y_train_c)
y_pred_dt = dt_clf.predict(X_test_c)

def evaluate_classifier(y_true, y_pred, model_name):
  print(f'\n--- {model_name} Performance ---')
  print(f'Accuracy:  {accuracy_score(y_true, y_pred):.4f}')
  print(f'Precision: {precision_score(y_true, y_pred, zero_division=0):.4f}')
  print(f'Recall:    {recall_score(y_true, y_pred, zero_division=0):.4f}')


evaluate_classifier(y_test_c, y_pred_log, 'Logistic Regression')
evaluate_classifier(y_test_c, y_pred_dt, 'Decision Tree Classifier')


plt.figure(figsize=(5, 4))
sns.heatmap(
    confusion_matrix(y_test_c, y_pred_log), annot=True, fmt='d', cmap='Blues'
)
plt.title('Confusion Matrix - Logistic Regression (Pass/Fail)')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

plt.figure(figsize=(6, 5))
plt.scatter(y_test, lir_pred, color='blue', alpha=0.6)
plt.xlabel('Actual Final Marks (G3)')
plt.ylabel('Predicted Marks')
plt.title('Actual vs Predicted Marks (Linear Regression)')
plt.plot([0, 20], [0, 20], color='red', linestyle='--')
plt.show()

```
